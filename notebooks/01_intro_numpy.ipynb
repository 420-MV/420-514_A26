# 1. Pourquoi utiliser NumPy ?

Une liste Python peut contenir des objets de types différents. NumPy est plutôt conçu pour manipuler de grandes quantités de données **homogènes**, c’est-à-dire généralement de même type.

Cela permet :

- un stockage plus compact en mémoire ;
- des calculs plus rapides ;
- des opérations vectorisées ;
- une syntaxe adaptée aux calculs scientifiques.

## 1.1 Liste Python vs tableau NumPy

````python
liste = [1, 2, 3, 4]
tableau = np.array([1, 2, 3, 4])

print(type(liste))
print(type(tableau))
````

Avec une liste Python, on utilise souvent une boucle ou une compréhension de liste pour appliquer une opération à tous les éléments.

````python
resultat_python = [x**3 for x in range(10_000)]
resultat_python[:5]
````

Avec NumPy, l'opération peut être appliquée directement au tableau :

```python
tableau ** 3
```

On parle d’**opération vectorisée**.

````python
d_numpy = np.arange(10_000)
resultat_numpy = d_numpy ** 3
resultat_numpy[:5]
````

### Mesurer les performances dans Jupyter

`%timeit` est une commande d’IPython/Jupyter qui exécute plusieurs fois une instruction afin d’estimer son temps d’exécution.

Les « loops » affichées par `%timeit` correspondent aux **répétitions du test de performance**, et non à des boucles Python utilisées par NumPy.

````python
%timeit [i**3 for i in range(10_000)]
````

````python
%timeit d_numpy ** 3
````

# 5. Calculs vectorisés et broadcasting

## 5.1 Opérations avec un scalaire

````python
a = np.array([1, 2, 3])

print(a + 2)
print(a * 3)
print(a ** 2)
````

Le scalaire est appliqué à chaque élément du tableau sans écrire explicitement une boucle Python.

## 5.2 Opérations élément par élément

````python
x = np.array([10, 20, 30])
y = np.array([2, 4, 5])

print("Addition       :", x + y)
print("Soustraction  :", x - y)
print("Multiplication:", x * y)
print("Division      :", x / y)
print("Puissance     :", x ** y)
````

Ces opérations sont effectuées **élément par élément** lorsque les formes sont compatibles.

## 5.3 Broadcasting

Le **broadcasting** permet à NumPy d’effectuer des opérations entre des tableaux de formes différentes lorsqu’elles sont compatibles.

Exemple : ajouter un vecteur de trois valeurs à chaque ligne d’une matrice de trois colonnes.

````python
matrice = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

ajustement = np.array([1, 2, 3])

matrice + ajustement
````

Dans cet exemple :

- `matrice.shape` vaut `(2, 3)` ;
- `ajustement.shape` vaut `(3,)`.

NumPy « étend » conceptuellement le vecteur sur les deux lignes.

# 9. Vues et copies

C’est une notion importante avec NumPy.

Une simple affectation peut faire référence aux **mêmes données en mémoire**.

````python
a = np.array([2, 4, 6, 8])
a1 = a

a1[0] = 99

print("a  :", a)
print("a1 :", a1)
print("Mémoire partagée :", np.shares_memory(a, a1))
````

### Créer une véritable copie

````python
a = np.array([2, 4, 6, 8])
a2 = a.copy()

a2[0] = 100

print("a  :", a)
print("a2 :", a2)
print("Mémoire partagée :", np.shares_memory(a, a2))
````

> **À retenir :** si vous devez modifier un tableau sans risquer de modifier l’original, utilisez `copy()`.

# 12. Agrégations et statistiques

````python
c = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
c
````

````python
print("Somme     :", np.sum(c))
print("Moyenne   :", np.mean(c))
print("Médiane   :", np.median(c))
print("Écart-type:", np.std(c))
print("Variance  :", np.var(c))
print("Minimum   :", np.min(c))
print("Maximum   :", np.max(c))
````

## 12.1 Comprendre `axis`

````python
print("Somme par colonne :", np.sum(c, axis=0))
print("Somme par ligne   :", np.sum(c, axis=1))
````

Pour une matrice :

- `axis=0` réduit les **lignes** et produit un résultat par **colonne** ;
- `axis=1` réduit les **colonnes** et produit un résultat par **ligne**.

````python
print("Médiane par colonne :", np.median(c, axis=0))
print("Médiane par ligne   :", np.median(c, axis=1))
````

## 12.2 Position du minimum ou du maximum

````python
print("Indice du maximum :", np.argmax(c))
print("Indice du minimum :", np.argmin(c))
````

## 12.3 Percentiles

````python
np.percentile(c, [25, 50, 75])
````

# 13. Valeurs manquantes avec `NaN`

Pour des données numériques réelles, une valeur absente est souvent représentée par `np.nan`.

`NaN` signifie *Not a Number*.

````python
mesures = np.array([12.5, 15.2, np.nan, 18.7, 16.4])

mesures
````

````python
np.isnan(mesures)
````

Les fonctions statistiques ordinaires propagent généralement `NaN`.

````python
np.mean(mesures)
````

NumPy fournit des variantes commençant par `nan` qui ignorent les valeurs manquantes.

````python
print("Moyenne :", np.nanmean(mesures))
print("Minimum :", np.nanmin(mesures))
print("Maximum :", np.nanmax(mesures))
````

# 17. Introduction à l’algèbre linéaire

## 17.1 Produit matriciel

````python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [2, 0],
    [1, 2]
])

A @ B
````

`A * B` effectue une multiplication **élément par élément**, alors que `A @ B` effectue un **produit matriciel**.

````python
print("Élément par élément :\n", A * B)
print("\nProduit matriciel :\n", A @ B)
````

## 17.2 Déterminant

````python
np.linalg.det(A)
````

## 17.3 Inverse

````python
np.linalg.inv(A)
````

## 17.4 Résoudre un système linéaire

````python
# 2x + y = 5
# x - y = 1

coefficients = np.array([
    [2, 1],
    [1, -1]
])

resultats = np.array([5, 1])

solution = np.linalg.solve(coefficients, resultats)
solution
````

# 19. Synthèse
